```python
import requests
import pandas as pd
```

```python
url1 = "https://api.worldbank.org/v2/country/all/indicator/NY.GDP.PCAP.CD?format=json&per_page=20000&date=2015:2019"
respuesta1 = requests.get(url1)

url2 = "https://api.worldbank.org/v2/country/all/indicator/SL.UEM.TOTL.ZS?format=json&per_page=20000&date=2015:2019"
respuesta2 = requests.get(url2)

url3 = "https://api.worldbank.org/v2/country/all/indicator/SP.URB.TOTL.IN.ZS?format=json&per_page=20000&date=2015:2019"
respuesta3 = requests.get(url3)
```

```python
pib = respuesta1.json()
pib[1][0]
```

```python
lista_pib = []

for registro in pib[1]:
    fila = {
        "country": registro["country"]["value"],
        "country_code": registro["countryiso3code"],
        "year": registro["date"],
        "gdp_per_capita": registro["value"]
    }
    lista_pib.append(fila)
```

```python
df_pib = pd.DataFrame(lista_pib)
df_pib.head(20)
```
