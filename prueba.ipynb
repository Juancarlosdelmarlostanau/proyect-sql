```python
import requests
import pandas as pd
```

```python
url1 = "https://api.worldbank.org/v2/country/all/indicator/NY.GDP.PCAP.PP.CD?format=json&per_page=20000&date=2015:2025"
api = requests.get(url1)
```

```python
api.status_code
```

```python
def limpiar_respuesta_api(api):
    datos_json = api.json()
    registros = datos_json[1]
    df = pd.DataFrame(registros)

    df['indicator_id'] = df['indicator'].apply(lambda x: x['id'] if isinstance(x, dict) else None)
    df['country_id'] = df['country'].apply(lambda x: x['id'] if isinstance(x, dict) else None)
    df['country_value'] = df['country'].apply(lambda x: x['value'] if isinstance(x, dict) else None)

    df = df.drop(columns=['indicator', 'country', 'obs_status', 'decimal'])
    columnas_ordenadas = ['country_id', 'country_value', 'date', 'value', 'indicator_id']
    df = df[columnas_ordenadas]

    df = df.rename(columns={
        'country_id': 'id_pais',
        'country_value': 'pais',
        'date': 'año',
        'value': 'gdp_per_capita',
        'indicator_id': 'id_indicador'})
    return df
```

```python
df_pib = limpiar_respuesta_api(api)
df_pib.head()
```

```python
def filtrar_por_años(df_pib):
    df_pib['año'] = df_pib['año'].astype(str)
    df_filtrado = df_pib[df_pib['año'].isin(['2015', '2024'])]

    return df_filtrado
```

```python
df_pib_filtrado = filtrar_por_años(df_pib)
df_pib_filtrado.head()
```
